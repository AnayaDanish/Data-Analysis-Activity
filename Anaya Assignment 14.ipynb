## DATA SET 1

```python
import pandas as pd
```

```python
df = pd.read_csv("pakistani_ecommerce_dataset.csv")
```

## Question 1: Most popular product category

```python
df.head()
```

```python
df['product_category'].value_counts()
```

### Home Decor is the most popular product category, with 30 orders. It is followed by Clothing with 22 orders, while Books has the lowest number of orders, with 11.

## Q2: Calculate the average order value

```python
df['order_value'].mean()
```

### The average order value is Rs. 2,832.15. This means that customers spend approximately Rs. 2,832 on average per order.

## Q3: City with the highest number of orders

```python
df['city'].value_counts()
```

### Karachi has the highest number of orders, with 24 orders. It is followed by Islamabad with 21 orders and Rawalpindi with 20 orders. Faisalabad has the lowest number of orders, with 17.

## Q4: Most commonly used payment method

```python
df['payment_method'].value_counts()
```

### Cash on Delivery is the most commonly used payment method, with 29 orders. It is followed by JazzCash with 27 orders, EasyPaisa with 23 orders, and Debit Card with 21 orders. This shows that Cash on Delivery is slightly more popular than the other payment methods.

## Q5: Bar chart of average order value by city

```python
city_avg = df.groupby('city')['order_value'].mean()
city_avg
```

### Islamabad has the highest average order value (Rs. 3,011.95), while Rawalpindi has the lowest (Rs. 2,447.08).

```python
import matplotlib.pyplot as plt
```

```python
city_avg.plot.bar()
plt.show()
```

### The bar chart shows the average order value across five cities. Islamabad has the highest average order value of Rs. 3,011.95, while Rawalpindi has the lowest at Rs. 2,447.08. Karachi and Faisalabad have relatively similar average order values."

## DATA SET 2

```python
import pandas as pd
```

```python
df2 = pd.read_csv("student_lifestyle_dataset.csv")
```

```python
df2.head()
```

## Q1: Calculate the average study hours

```python
df2['study_hours'].mean()
```

### The average study time of students is approximately 4.40 hours per day. This means that students spend around 4 hours and 24 minutes studying daily on average.

## Q2: Average sleep hours by gender

```python
df2.groupby('gender')['sleep_hours'].mean()
```

### Female students sleep for an average of 6.80 hours, while male students sleep for an average of 6.39 hours. This shows that female students in this dataset get slightly more sleep than male students.

## Q3: Relationship between screen time and sleep

```python
df2['screen_time'].corr(df2['sleep_hours'])
```

### The correlation between screen time and sleep hours is approximately 0.052. This indicates a very weak positive relationship between the two variables. In this dataset, screen time does not appear to have a strong relationship with the number of hours students sleep.

## Q4: Histogram of daily transport cost

```python
df2['daily_transport_cost'].plot.hist()
plt.show()
```

### The histogram shows the distribution of daily transport costs among students. The costs vary across different ranges, with the highest frequency appearing around Rs. 50–75. There is also a noticeable concentration around Rs. 275–300.

## Question 5: Average internet usage by gender

```python
df2.groupby('gender')['internet_usage'].mean()
```

### Male students have an average internet usage of 6.00 hours, while female students have an average of 5.32 hours. This shows that male students spend slightly more time using the internet than female students in this dataset.
