```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_tea_snacks.csv")
```

```python
df. head()
```

## Q1: What is the most preferred tea type?

```python
df["tea_type"].value_counts()
```

```python
df["tea_type"].value_counts().idxmax()
```

### The most preferred tea type among students is the one with the highest number of responses, i.e is green.

## Q2: What is the average number of cups per gender?

```python
df.groupby("gender")["cups_per_day"].mean()
```

### This compares the average number of tea cups consumed by male and female students.

## Q3: Make a bar chart of snack choices.

```python
df.columns
```

```python
df["snack_choice"].value_counts().plot(kind="bar")
plt.show()
```

### The bar chart shows the most, i.e samosas and least i.e biscuits as preferred snacks among students.

## Q4: Compare tea preference by age group.

```python
df.groupby("age_group")["tea_type"].value_counts()
```

### This shows how tea preferences differ across different age groups.

## Q5: Count the number of students who don't drink tea.

```python
(df["tea_type"] == "None").sum()
```
